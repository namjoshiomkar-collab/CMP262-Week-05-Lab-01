## Part 7 - Final EDA Challenge

Use pandas to answer this question: **Which delivery mode has the higher average satisfaction score?**

1. Calculate the average `Satisfaction` separately for `Online` and `In Person` workshops.
2. Display both averages.
3. In the markdown cell below, write 2-3 sentences explaining your conclusion and mention the values that support it.


````python
# Filter for online workshops
online_workshops = workshops[workshops["Delivery_Mode"] == "Online"]
online_satisfaction_mean = online_workshops["Satisfaction"].mean()

# Filter for in-person workshops
in_person_workshops = workshops[workshops["Delivery_Mode"] == "In-Person"]
in_person_satisfaction_mean = in_person_workshops["Satisfaction"].mean()

# Print the results
print("Online Satisfaction Mean:", in_person_satisfaction_mean)
print("In-Person Satisfaction Mean:", online_satisfaction_mean)
````

Write your conclusion here.


The workshops filter the delivery mode to be either online or in person. Then we get the satisfaction average of that variable.
